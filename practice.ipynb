```python
measurements = [(1, 2.1), (2, 3.8), (3, 6.2), (4, 7.9), (5, 10.1)]

# Assume shake is roughly proportional to speed
# Estimate the rate by averaging the observed shake per unit of speed. 

shake_per_unit_speed = sum(
    shake / speed for speed, shake in measurements
) / len(measurements)

for speed in (0, 3.5, 8):
    prediction = shake_per_unit_speed * speed
    print(f"at {speed} m/s, predicted shake is {prediction:.2f} mm ")
```

```python
test_speed = 3.5 
actual_shake = 7.4 

predicted_shake = shake_per_unit_speed * test_speed
absolute_error = abs(actual_shake - predicted_shake)

print(f"Predicted shake : {predicted_shake:.2f} mm")
print(f"Actual shake : {actual_shake:.2f} mm")
print(f"Absolute error : {absolute_error:.2f} mm")
```

```python
test_measurements = [(1.5, 3.0), (3.5, 7.4), (4.5, 8.5)]

squared_errors = []

for speed, actual_shake in test_measurements:
    predicted_shake = shake_per_unit_speed * speed
    error = actual_shake - predicted_shake
    squared_error = error ** 2
    squared_errors.append(squared_error)

    print(f"{speed} m/s | error: {error:.2f} mm | squared error: {squared_error:.2f} mm^2")

mse = sum(squared_errors) / len(squared_errors)
print(f"Mean squared error: {mse:.2f} mm^2")
```

```python
training_runs = [(1, 2.1), (2, 3.8), (3, 6.2), (4, 7.9), (5, 10.1)]
test_runs = [(1.5, 3.0), (3.5, 7.4), (4.5, 8.5)]

def mse(runs, slope):
    squared_errors = []

    for speed, actual_shake in runs:
        predicted_shake = slope * speed
        error = actual_shake - predicted_shake
        squared_errors.append(error ** 2)

    return sum(squared_errors) / len(squared_errors)

# Training: compare two candidate slopts using only the training runs. 
print("Training MSE with slope 1.5:", mse(training_runs, 1.5))
print("Training MSE with slope 2.0:", mse(training_runs, 2.0))
```
